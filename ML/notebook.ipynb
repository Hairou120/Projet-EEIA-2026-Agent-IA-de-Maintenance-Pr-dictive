#  Agent IA de Maintenance Prédictive

## Contexte

Dans les environnements industriels, les pannes machines imprévues 
entraînent des arrêts de production coûteux et des pertes importantes. 
La maintenance corrective (réparer après la panne) est inefficace. 
La maintenance prédictive permet d'anticiper les défaillances 
avant qu'elles ne surviennent, grâce à l'analyse des données capteurs.

## Objectif

Construire un modèle de Machine Learning capable de prédire 
si une machine industrielle va tomber en panne, à partir de 
ses données de fonctionnement en temps réel (température, 
vitesse, couple, usure).

## Dataset

- Source : AI4I 2020 Predictive Maintenance Dataset
- Taille : 10 000 observations, 14 variables
- Variable cible : Machine failure (0 = normal, 1 = panne)

## Approche

Nous construisons ce projet de façon rigoureuse et progressive :
exploration → nettoyage → préparation → modélisation → évaluation → interprétation.

Chaque décision technique sera justifiée.

## Description des variables

### Variables d'identification (nous allons revoir sa ensemble)

| Variable | Signification | Type | Rôle |
|---|---|---|---|
| UDI | Identifiant unique de la ligne | Entier | Aucun → à supprimer |
| Product ID | Code produit de la machine | Texte | Aucun → à supprimer |

### Variables explicatives (nos X)

| Variable | Signification | Type | Unité |
|---|---|---|---|
| Type | Qualité de la machine | Catégorielle | L=Low, M=Medium, H=High |
| Air temperature | Température de l'air ambiant | Numérique | Kelvin (K) |
| Process temperature | Température interne du processus | Numérique | Kelvin (K) |
| Rotational speed | Vitesse de rotation | Numérique | Tours/minute (RPM) |
| Torque | Couple moteur | Numérique | Newton-mètre (Nm) |
| Tool wear | Durée d'usure de l'outil | Numérique | Minutes |

### Variable cible (notre Y)

| Variable | Signification | Type | Valeurs |
|---|---|---|---|
| Machine failure | La machine a-t-elle eu une panne ? | Binaire | 0=Normal, 1=Panne |

### Types de pannes (informations complémentaires)

| Variable | Signification |
|---|---|
| TWF | Panne par usure de l'outil |
| HDF | Panne par dissipation thermique |
| PWF | Panne électrique |
| OSF | Panne par surcharge |
| RNF | Panne aléatoire |

> information capitale : Les types de pannes (TWF, HDF, PWF, OSF, RNF) 
> sont des sous-catégories de Machine failure. 
> Nous devrons faire attention à ne pas les utiliser 
> comme variables X car elles causeraient une fuite de données.
